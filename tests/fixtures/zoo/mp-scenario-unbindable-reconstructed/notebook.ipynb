## 3. The setup pieces


```
# The scene hides behind an opaque world object: no obstacle-named
# list and no `.obstacles` attribute for a detector to bind.
class _World:
    def __init__(self):
        self._shapes = [("circle", 1.5, 1.5, 0.4)]

    def blocked(self, x, y):
        return any(((x - sx) ** 2 + (y - sy) ** 2) ** 0.5 < r
                   for _, sx, sy, r in self._shapes)


world = _World()
start = [0.0, 0.0, 0.0]
goal = [5.0, 5.0]
```
